#### Scatter Plot(산점도)
    : 두 변수의 관계를 2차원 평면위에 점으로 나타내는 그래프
    - 상관관계 파악
        - 양의 상관관계 : x가 증가하면, y도 증가하는 경향
        - 음의 상관관계 : x가 증가하면, y는 감소하는 경향
        - 무상관 : x와 y 사이의 특별한 선형 관계가 없음

    - 이상치 감지를 한눈에 파악할 수 있다
    - 점들이 특정 영역에 뭉쳐(군집(cluster)) 있는지 확인이 쉽다

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.figure(figsize=(8,4))
markers = ['.', ',','1','_','>','<','^','2','3','4']

for m in markers:
    plt.plot(np.random.rand(5), np.random.rand(5), m, label="'{0}'".format(m))

plt.legend(loc='center right', ncol=2)
plt.xlim(0,1.5)
```

```python
x = np.linspace(0,10,100)
y = np.sin(x)

plt.scatter(x,y, marker='*')
```

```python
for i in range(9):
    x = np.arange(1000)
    y = np.random.randn(1000).cumsum()
    plt.scatter(x, y, alpha=0.4, c=y,cmap = 'tab20c')
```

#### 버블 차트

```python
x = np.random.randn(100)
y = np.random.randn(100)
colors = np.random.rand(100)
sizes = np.random.rand(100)*1000

plt.scatter(x,y, c=colors, s=sizes, alpha=0.5, cmap='tab20c')
plt.colorbar()
```

```python
x = np.random.randn(200)
y1 = np.random.randn(len(x))
y2 = 1.1 * np.exp(x)

ax1 = plt.plot()
plt.scatter(x, y1, color='navy', alpha= 0.3, label='no correlation')
plt.scatter(x, y2, color='black', alpha= 0.3, label='correlation')
plt.grid(True)
plt.legend()
```
