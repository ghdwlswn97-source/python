```python
df6 = DataFrame({'addr' : ['충청','제주','전라'],
                'tel' : ['042', '051','076'],
                'department' : ['컴퓨터공학', '통계학과', '디자인학과']
                })

df6
```

```python
import pandas as pd
```

```python
# on = 'addr' 이면 addr이 겹치는 것만 합치기
result = pd.merge(df2, df6, on = 'addr')
result
```

```python
result2 = pd.merge(df2, df6, how='left')        # df2를 기준으로 합쳐짐
result2
```

```python
result3 = pd.merge(df2, df6, how='right')       # df6을 기준으로 합쳐짐
result3
```

```python
df7 = DataFrame({'addr' : ['충청','제주','경기'],
                'tel' : ['042', '051','076'],
                'department' : ['컴퓨터공학', '통계학과', '디자인학과']
                })
```

```python
result4 = pd.merge(df2, df7, how='outer')       # 한쪽 데이터프레임에만 있는 값도 합쳐짐
result4
```

```python
# 데이터 프레임 연결
result5 = pd.concat([df2,df7])
result5
```
