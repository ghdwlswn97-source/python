```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.use(['seaborn-v0_8-deep','seaborn-v0_8-notebook'])
# plt.rcParams.update({
#     'axes.facefolor' : 'white'
#     'grid.color' : 'gray'
#     'grid.alpha': 0.5
# })

data = np.random.randn(10000)

plt.hist(data, edgecolor='black')
```

```python
plt.hist(data, bins=30, color = 'purple', edgecolor="#2B0463", linewidth=1.2, alpha=0.5)
```

```python
x1=np.random.normal(0,1,10000)
x2 = np.random.normal(-5,3,10000)
x3=np.random.normal(5,2,10000)
d= dict(histtype='stepfilled', alpha=0.3, bins=50)

plt.hist(x1, **d)
plt.hist(x2, **d)
plt.hist(x3, **d)
```

```python
import matplotlib.colors as colors
x=np.random.normal(size=50000)
y=x-np.random.normal(size=50000)

plt.hist2d(x,y, bins=50, cmap='OrRd', norm=colors.LogNorm())
plt.colorbar()
```
