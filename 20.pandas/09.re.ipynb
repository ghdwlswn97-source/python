#### Regular Expression
: 파이썬 정규표현식(Regex)을 다루는 내장 라이브러리

```python
import re
```

```python
# one문자가 있는지 확인
pattern = 'one'
str1 = 'one12345two'
re.match(pattern, str1)
```

```python
match1 = re.match(pattern, str1)
match1.group()
```

```python
p = re.compile(r'(\w+)\s+\d+-\d+')
```

```python
result = p.search('tjoeun 02-1234-1234')
result.group(1)
```

```python
p = r'(\w+)\s+(\d+)-(\d+)-(\d+)'
text = 'tjoeun 02-1234-1234'
print(re.findall(p, text))


p = r'\w+\s+\d+-\d+-\d+'
text = 'tjoeun 02-1234-1234'
print(re.findall(p, text))
```

```python
source1 = 'modelA 공급면적 220.98제곱미터 전용면적 193.43제곱미터'

result1 = re.findall(r'공급면적\s+(.*?)제곱미터', source1)
print('공급면적 :', result1)

result2 = re.findall(r'전용면적\s+(.*?)제곱미터', source1)
print('전용면적 :', result2)
```

```python
source_a = 'dodo/mimi//solsol///lala,sisi'
```
