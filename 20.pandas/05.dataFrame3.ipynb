```python
from pandas import DataFrame
```

```python
df = DataFrame({'거래처':['company1','company2','company3','company4','company5'],
                '주문량':[1500, 3040, 3389, 3343,1283],
                '할인율':[0.2,0.3,0.16,0.31,0.05]
                })
df
```

```python
# 합계 : 문자는 나열
df.sum()
```

```python
# 누적합계
df.cumsum()
```

```python
# 나누기 -> 문자는 나누기 불가로 오류
# df.mean()

# numeric_only=True 옵션은 숫자만 평균 내기
df.mean(numeric_only=True)
```

```python
# 요약 정리
df.describe()
```

```python
df.sum(axis=1, numeric_only=True)
# df.sum(1, numeric_only=True)      axis 생략 가능
```
