```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [10,40,30,60,20]
categories=['a1', 'a2', 'a3','a4','a5']

# explode -> 원형 그래프(pie chart)에서 특정 조각을 바깥쪽으로 분리해서 보여주는 옵션
plo =[0.1 for _ in range(5)]
plt.pie(data, explode = plo, startangle=90, labels=categories, autopct='%.1f%%',
        colors=['navy','green','blue','red','black'])      # %% -> 숫자뒤에 % 표시
plt.legend(categories)
# startangle=90 90도 회전
```

```python
plt.figure(figsize=(6,6))

plt.pie(data, labels= categories,autopct='%.1f%%',startangle=90 )

# 중앙에 흰색 원 추가
center_circle=plt.Circle((0,0), 0.4, fc='white')
fig=plt.gcf()
fig.gca().add_artist(center_circle)
plt.title('2D Doughnut Chart')
```
