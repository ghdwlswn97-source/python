#### 문제 7. 그룹별 데이터 가져오기

문제 6에서 불러와서 저장한 변수의 값들을 선별,
열 기준으로 그룹화한 다음, '경부선' 그룹만 df_group1 변수에 저장하고 출력

```python
df_group1 = df_excel.groupby('선별').count()
df_group1=df_excel[df_excel['선별'] == '경부선']
df_group1
```

#### 문제 8. 조건 두 개를 동시에 만족하는 행 찾기

문제 6에서 불러와서 저장한 변수의 값들을 선별이 '경부선'이고 총이용인원이 1000 이상인 행만 추출

```python
df_result=df_excel[(df_excel.선별=='경부선')&(df_excel.총이용인원 >= 1000)]
df_result
```

#### 문제 9. 그룹별 개수·평균·합계 구하기

2. 각 그룹의 숫자 열 평균
3. 각 그룹의 숫자 열 합계

```python
print(df_excel.groupby('선별').count())
print(df_excel.mean(numeric_only=True))
print(df_excel.sum())
print(df_excel['차종'].count())
```
