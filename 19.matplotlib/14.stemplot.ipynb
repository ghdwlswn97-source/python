#### Stem plot(줄기-싶 그래프)
    - 데이터의 연속적인 값과 개별측정값을 보존하면서 전체적인 분포 모양을 보여주는 그래프
    - 특징
        - 연속/신호 데이터 시각화 : 이산신호 처리나 시간 흐름에 따른 데이터 이산점 표시에 쓰임
        - 장점 : 개별 데이터 포인트의 위치와 크기를 명확히 파악가능

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0.1, 2*np.pi, 41)
y = np.exp(np.sin(x))

plt.stem(x,y, linefmt='gray', markerfmt='b', basefmt='g', bottom=1)       # linefmt : 라인의 색상
```

#### Box plot
    - 데이터의 통계량(5수치 요약 : 1사분위수, 중앙값, 3사분위수, 최대값)을 상자 형태
    - 구조
        - 상자(Box) : 데이터의 중앙 50%범위를 나타냄
        - 중앙값 선 : 상자 내부에 선으로 데이터의 가운데값 표시
        - 수염(whisker) : 정상 데이터의 최대값과 최소값까지 늘어난 선
        - 점 : 수염의 범위를 벗어난 극단값을 이상치 점으로 표시
    - 장점 : 여러 그룹간의 데이터 분포, 편향성, 이상치 존재여부 비교에 용이
     

```python
r1=np.random.normal(loc=0, scale=0.5, size=100)
r2=np.random.normal(loc=0.5, scale=1, size=100)
r3=np.random.normal(loc=1, scale=1.5, size=100)
r4=np.random.normal(loc=1.5, scale=2, size=100)
r5=np.random.normal(loc=2, scale=2.5, size=100)

f, ax = plt.subplots(1,1)
ax.boxplot((r1, r2, r3, r4, r5))
ax.set_xticklabels(['r1', 'r2', 'r3', 'r4', 'r5'])
```
