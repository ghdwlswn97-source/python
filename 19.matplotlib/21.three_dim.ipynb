```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')
```

```python
x =range(1, 101)
y = np.random.randn(100) * x
z = np.random.randn(100) * x

fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')
ax.scatter(x, y, z, c='green', s=60)
```

```python
fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')

zline = np.linspace(0, 20, 1000)
xline = np.sin(zline)
yline = np.cos(zline)
ax.plot3D(xline, yline, zline, 'gray')

zdata = 20 * np.random.random(100)
xdata = np.sin(zdata) + 0.2 * np.random.randn(100)
ydata = np.cos(zdata) + 0.2 * np.random.randn(100)
ax.scatter3D(xdata, ydata, zdata, c=zdata, cmap='Blues')
```

```python
m = np.linspace(-4, 4, 20)
x, y = np.meshgrid(m, m)
z = np.cos(np.sqrt(x**2 + y**2))

fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')

ax.contour3D(x, y, z, 50, cmap='BuPu')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
```

```python
ax.view_init(70, 50)
fig
```

```python
fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')
ax.plot_wireframe(x,y,z, color='blue');
```

```python
fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')
ax.plot_surface(x,y,z, rstride=1, cstride=1, cmap='viridis');
```

```python
r = np.linspace(0, 6, 20)
theta = np.linspace(-0.8 * np.pi, 0.8 * np.pi, 40)
r, theta = np.meshgrid(r, theta)

x = r * np.sin(theta)
y = r * np.cos(theta)
z = np.cos(np.sqrt(x**2 + y**2))

fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')
ax.plot_surface(x, y, z, cmap='viridis')
ax.set_box_aspect([1.5, 1, 1])
```

```python
theta = 2 * np.pi * np.random.random(1000)
r = 6 * np.random.random(1000)
x = np.ravel(r * np.sin(theta))
y = np.ravel(r * np.cos(theta))
z = np.cos(np.sqrt(x**2 + y**2))

fig = plt.figure(figsize=(6,6))
ax = plt.axes(projection='3d')
ax.scatter(x, y, z, c=z, cmap='viridis')
```

```python
a = np.linspace(-1, 1, 100)
b = np.linspace(-1, 1, 100)
A, B = np.meshgrid(a,b)
L = A**2 + B**2
plt.contour(a,b,L, 20)
plt.xlabel('a')
plt.ylabel('b')
plt.grid()

plt.figure()
plt.contourf(a,b,L, 5)
```
