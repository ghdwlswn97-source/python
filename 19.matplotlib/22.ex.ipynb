문제 1. 한 가게의 상품별 판매량 데이터를 사용해 막대그래프와 원그래프를 각각 그리기
products = ['Coffee', 'Tea', 'Juice', 'Water']
sales = [120, 80, 60, 40]

* 요구사항
1. 그래프 2개를 나란히 배치
2. 첫 번째 그래프는 bar plot으로 상품별 판매량을 표시.
3. 막대 색상은 상품마다 다르게 지정하고, 제목과 축 이름을 추가
4. 두 번째 그래프는 pie chart로 판매 비율을 표시
5. 원그래프에 퍼센트(autopct)와 상품 이름을 표시

```python
products = ['Coffee', 'Tea', 'Juice', 'Water']
sales = [120, 80, 60, 40]

plt.subplot(1,2,1)
plt.bar(products, sales)
plt.title('products_sales')


plt.subplot(1,2,2)
plt.pie(sales, labels=sales, autopct='%.1f%%',colors=['navy','green','blue','red'])
```
