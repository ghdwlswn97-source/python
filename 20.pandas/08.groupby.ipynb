```python
df_result=df[(df.선별== '경부선 ')&(df.총이용인원 >=1000)]
df_result
```

```python
df_result.sort_values(by='총이용인원', ascending=0).head()
```

```python
# groupby() 뒤에는 반드시 집계함수가 있어야 한다
# sum(), mean(), count(), max() .....
df.groupby('선별').count()
```

```python
df_group = df.groupby('선별')
df_group['차종'].count().sort_values(ascending=0)
```

```python
import matplotlib.pyplot as plt

plt.rc('font', family='Malgun Gothic')

df_group['차종'].count().plot(kind='bar')
plt.show()
```

```python
df_avg = df.groupby(['차종','선별'])
df_avg['총이용인원'].mean()
df_avg[['총이용인원', '이용율']].mean()

# 한줄로 정리
df.groupby(['차종','선별'])[['총이용인원', '이용율']].mean()
```

```python
pd.crosstab(df['차종'], df['선별'])
pd.crosstab(df['차종'], df['선별'], margins=True)       # margins=True -> 총합
```
