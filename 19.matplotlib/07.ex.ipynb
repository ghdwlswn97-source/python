```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig=plt.figure()
plt.axis()
```

```python

day= [1,2,3,4,5]
study=[2,3,4,5,6]
plt.plot(day,study, 'b-o')
plt.title('Study Hours')
plt.xlabel('Day')
plt.ylabel('Hours')
```

```python
day = [1, 2, 3, 4, 5, 6, 7]
temp = [20, 22, 21, 25, 27, 26, 29]
plt.plot(day,temp, 'r--s', linewidth=2,markersize=8)
# plt.xlim(1,7)
# plt.ylim(15,35)
plt.axis([1,7,15,35])
plt.title('Weekly Temperature')
```

```python
exam = [1, 2, 3, 4, 5]
A = [70, 75, 80, 85, 90] 
B = [65, 72, 78, 82, 88]

plt.plot(exam, A, 'b-o')
plt.plot(exam,B, 'g--^')
plt.title('Exam Score Comparison')
plt.xlabel('Exam')
plt.ylabel('Score')
plt.legend('Student A, Student B', loc= 'lower right')
```

```python

month = [1, 2, 3, 4, 5, 6]
sales = [120, 150, 130, 180, 200, 230]

plt.rcParams['font.family']='Malgun Gothic'
plt.rcParams['axes.unicode_minus']= False

plt.plot(month, sales, '-.D', color='purple' )
plt.xlim(1,6)
plt.ylim(100,250)
plt.title('월별 매출변화', fontsize=16)
plt.xlabel('월',fontsize=12)
plt.ylabel('판매',fontsize=12)
plt.legend('매출액')
```

```python
month = [1, 2, 3, 4, 5, 6]

A=product_A = [100, 120, 140, 130, 160, 180]
B = product_B = [90, 110, 115, 140, 150, 170]
C=product_C = [80, 95, 105, 120, 135, 150]

plt.plot(month,A,'r-o', linewidth=2)
plt.plot(month,B, 'b--s',linewidth=2)
plt.plot(month,C, 'g:^',linewidth=2)

plt.xlim(1,6)
plt.ylim(50,200)
plt.xlabel('Month')
plt.ylabel('Sales')

plt.legend('upper left')

plt.title('Product Sales Comparison', size=18)

plt.xticks([1,2,3,4,5,6])


```
