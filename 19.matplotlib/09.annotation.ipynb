#### 텍스트와 주석

- ax.transData(데이터 좌표계)<br>
    - 기준 : 그래프의 데이터 축 값(x축, y축 눈금)
    - ax.axis([0,10,0,10])으로 설정한 데이터 눈금을 그대로 따른다<br>
        x=3, y=6 위치에 데이터 표시

- ax.transAxes(축 영역 좌표계)
    - 기준 : 개별 subplot(Axes)의 상대적 배율(0.0 ~ 1.0)
    - subplot의 좌측 하단이 (0,0), 우측 상단이 (1,1)이 됨<br>
        (0.2,0.4) => 가로의 20%, 세로의 40% 지점에 위치

- fig.transFigure(전체 그림 좌표계)
    - 기준 : 전체 캔버스의 상대적 비율(0.0 ~1.0)
    - 여백을 포함한 전체 이미지 창의 좌측 하단이 (0,0), 우측 상단이 (1,1)이 됨<br>


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig, ax = plt.subplots()

ax.axis([0,10,0,10])
ax.text(3,6, '. transData(3,6)', transform=ax.transData)        # 그래프 눈금 (3, 6) 위치
ax.text(0.2, 0.4, '. transAxes(0.2, 0.4)', transform=ax.transAxes)      # 그래프의 20%, 40% 위치
ax.text(0.2, 0.2, '. transFigure(0.2, 0.2)', transform=fig.transFigure)     # 여백을 포함한 전체 그림의 20%, 20% 위치

```

```python
ax.set_xlim(-6,10)
ax.set_ylim(-6,10)
fig
```

```python
x=np.arange(1,40)
y=x*1.1
plt.scatter(x,y,marker='.')
plt.axis('equal')
plt.annotate('interesting point', xy=(4,5), xytext=(20,10), arrowprops=dict(shrink=0.1))
```

- 화살표 style

|종류|설명|
|:---:|:---:|
|-|머리가 없는 화살표|
|->|기본 얇은 화살표|
|-\|>|머리가 채워진 화살표|
|<->|양방향 화살표|
|fancy|머리와 줄기가 굵은 곡선감이 있는 화살표|
|simple|심플한 삼각형 머리의 굵은 화살표|
|wedge|부채꼴 모양의 화살표|

```python
x1=np.random.normal(30,3,100)
x2=np.random.normal(20,3,100)
x3=np.random.normal(10,3,100)

plt.plot(x1, label='p1')
plt.plot(x2, label='p2')
plt.plot(x3, label='p3')

plt.legend(bbox_to_anchor=(0., 1.0, 1.,0.1), loc=0, ncol=3, mode='extend', borderaxespad=3.)
plt.annotate('important value', (50, 20), xytext=(5,40), arrowprops=dict(arrowstyle='->'))
plt.annotate('important value', (40, 30), xytext=(50,40), arrowprops=dict(arrowstyle='->'))
```
