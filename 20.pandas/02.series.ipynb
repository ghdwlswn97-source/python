#### Series
: 인덱스가 붙어있는 1차원 데이터

```python
from pandas import Series
import numpy as np
```

```python
# data1 = Series([45, 87,14,59])
data1 = Series([45, 87,14,59, '더조은', True, 3.3837])      # 숫자, 문자, 실수, True, False 다 가능
data1
```

```python
data2 = Series(np.random.randint(1,10, size = (5,)), index = ['1회', '2회', '3회', '4회', '5회'])       # 인덱스 이름 설정
data2
```

```python
data2['2회']        # 특정 인덱스의 해당 값 가져오기
data2['2회'] = 9        # 특정 인덱스 값 변경
data2
```

```python
data2.index     # 인덱스만 가져오기
```

```python
data2.values        # 값만 가져오기
```

```python
age = {'kim' : 28, 'lee' : 32, 'park' : 25}
data3 = Series(age)
print(data3)

data4 = Series(age, index=['no1', 'no2','no3'])     # 키 : 값으로 들어오면 인덱스 수정 불가
data4
```
