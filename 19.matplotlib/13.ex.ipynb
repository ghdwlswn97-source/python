```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = [1, 2, 3, 4, 5, 6]
sales_a = [20, 27, 25, 35, 42, 48]
sales_b = [17, 21, 30, 29, 37, 44]

ax= plt.axes 
fig=plt.figure

fig, ax = plt.subplots(1,2, sharex=True, sharey=True)

ax1 = fig.add_subplot(1,2,1)
ax1.plot(months, sales_a, '-r', marker='o', label='A')
ax1.set_xlim(1,6)
ax1.set_ylim(0,55)
ax1.set_title('Product Sales Dashboard')
ax1.set_xlabel('m')
ax1.set_ylabel('a')
ax1.legend()

ax2 = fig.add_subplot(1,2,2)
ax2.plot(months, sales_b, '--b', marker='s', label='B')
ax2.set_xlim(1,6)
ax2.set_ylim(0,55)
ax2.set_title('Product Sales Dashboard')
ax2.set_xlabel('m')
ax2.set_ylabel('b')
ax2.legend()


```

```python
weeks = [1, 2, 3, 4, 5, 6]
scores = [65, 71, 78, 74, 89, 84]

plt.plot(weeks, scores, color='purple', marker='o')
plt.title('Weekly Test Scores.')
plt.xlim(1,6)
plt.ylim(50,100)
plt.annotate('Peak: 89', xy=(5,89), xytext=(3.5, 96), arrowprops=dict(arrowstyle='->'))
```

```python
days = [1, 2, 3, 4, 5, 6, 7]
city_a = [18, 20, 19, 23, 25, 24, 26]
city_b = [15, 17, 18, 20, 21, 23, 22]

plt.plot(days, city_a )
plt.plot(days, city_b)
plt.title('City Temperatures')
plt.xlabel('Day')
plt.ylabel('Temperature (C)')
plt.annotate('Weekly comparison', xy=(0.05, 0.9), xycoords=ax1.transAxes)
```

```python
quarters = [1, 2, 3, 4]
revenue = [120, 145, 138, 175]
cost = [80, 86, 92, 96]

fig, ax = plt.subplots(1,2, sharex=True, sharey=True)
fig.text(0.3,0, 'Source: Sample business data')
ax1=fig.add_subplot(2,1,1)
ax1.plot(quarters,revenue, label='A')
ax1.set_title('revenue')


ax2=fig.add_subplot(2,1,2)
ax2.plot(quarters,cost, label='B')
ax2.set_title('cost')

```
