##### 플롯 종류

| 함수 | 설명 |
| :---: | :---: |
| bar | 막대 그래프 |
| barbs | 기상학에서 사용하는 바람장(풍향 및 풍속) 기호(Wind Barbs)를 그리기 |
| boxplot | 상자 및 수염 그래프 |
| cohere | x와 y의 일관성 시각화 그리기 |
| contour | 등고선 |
| errorbar | 오류 막대 그래프 |
| hexbin | 육각형 binning 플롯 |
| hist | 히스토그램 플롯 |
| imshow | 축에 이미지 표시 |
| pcolor | 2차원 배열의 유사 플롯 생성 |
| pcolormesh | 사각 망사 그래프 |
| pie | 파이 챠트 플롯 |
| plot | 플롯 라인 또느 마커 |
| quiver | 화살표의 2차원 필드 생성 |
| sankey | Sankey 흐름도 생성 |
| scatter | x대 y의 산점도 생성 |
| stem | 줄기 그래프 생성 |
| streamplot | 벡터 흐름의 스트림 라인 그리기 |


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.style.use(['seaborn-v0_8-deep'])
```

```python
height = [np.random.randn()*i for i in range(1,6)]
names = ['A', 'B', 'C', 'D', 'E']
y_pos = np.arange(len(names))
plt.bar(y_pos, height)      # bar -> 세로 막대
plt.xticks(y_pos, names, fontweight='bold')
plt.xlabel('group')
```

```python
height = [np.random.randn()*i for i in range(1,6)]
names = ['A', 'B', 'C', 'D', 'E']
y_pos = np.arange(len(names))
plt.barh(y_pos, height)     #barh -> 가로 막대
plt.yticks(y_pos, names, fontweight='bold')
plt.ylabel('group')
```

```python
bar1=[4,25,3,12,7]
bar2=[8, 44, 23, 6, 7]
bar3=[3, 6, 4, 32, 7]
bars= np.add(bar1, bar2).tolist()       # bar1+bar2 한뒤 리스트로 만들기
print(bars)

r = [0,1,2,3,4]
names = ['A', 'B', 'C', 'D', 'E']

plt.bar(r, bar1, color='navy', edgecolor='white')
plt.bar(r, bar2, bottom=bar1, color='red', edgecolor='white')
plt.bar(r, bar3, bottom = bar2, color='green', edgecolor='white')

plt.xlabel('group', fontweight='bold')
plt.xticks(r, names, fontweight='bold')
```

```python
bar_width=0.25

bar1=[4,25,3,12,7]
bar2=[8, 44, 23, 6, 7]
bar3=[3, 6, 4, 32, 7]

r1=np.arange(len(bar1))
r2 = [x+bar_width for x in r1] 
r3= [x+bar_width for x in r2]

plt.bar(r1, bar1, color='olive', edgecolor = 'black', width= bar_width, label =r1)
plt.bar(r2, bar2, color='darkkhaki', edgecolor = 'black', width= bar_width, label =r2)
plt.bar(r3, bar3, color='green', edgecolor = 'black', width= bar_width, label =r3)

plt.xlabel('group', fontweight='bold')
plt.xticks([i + bar_width for i in range(len(bar1))], ['A', 'B', 'C', 'D', 'E'])
plt.legend()
```
