```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig = plt.figure(figsize=(15,15))
x = range(1,11)
colums = [np.random.randn(10)*i for i in range(1,26)]

for n, v in enumerate(plt.style.available[:25]):     
    plt.style.use(v)
    plt.subplot(5,5, n+1)
    plt.title(v)

    for c in colums:
        plt.plot(x, c, color='teal', linewidth=1, alpha=0.4)

plt.subplots_adjust(hspace=0.5, wspace=0.5)
```
