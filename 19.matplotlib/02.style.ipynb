#### Line Style
| 문자 | 스타일 |
| :---: | :---: |
| - | solid |
| -- | dashed |
| -. | dashdot |
| : | dotted

```python
import matplotlib.pyplot as plt
import numpy as np
# plt.style.use('seaborn-v0_8-deep')
```

```python
plt.plot(np.random.randn(50).cumsum(), linestyle= 'solid')
plt.plot(np.random.randn(50).cumsum(), linestyle= 'dashed')
plt.plot(np.random.randn(50).cumsum(), linestyle= 'dashdot')
plt.plot(np.random.randn(50).cumsum(), linestyle= 'dotted')
```

```python
plt.plot(np.random.randn(50).cumsum(), linestyle='-')
plt.plot(np.random.randn(50).cumsum(), linestyle = '--')
plt.plot(np.random.randn(50).cumsum(), linestyle = '-.')
plt.plot(np.random.randn(50).cumsum(), linestyle = ':')
```

- Color Style

| 문자 | 색상 | 
|:---:|:---:|
| b | blue |
| g | green |
| r | red |
| c | cyan |
| m | magenta |
| y | yellow |
| k | black |
| w | white |

```python
plt.plot(np.random.randn(50).cumsum(), color = 'g')
plt.plot(np.random.randn(50).cumsum(), color = '#5CD1E5')
plt.plot(np.random.randn(50).cumsum(), color = (0.2, 0.4, 0.6))
plt.plot(np.random.randn(50).cumsum(), color ='chocolate')
```

```python
plt.plot(np.random.randn(50).cumsum(), 'b-')
plt.plot(np.random.randn(50).cumsum(), 'g--')
plt.plot(np.random.randn(50).cumsum(), 'm-.')
plt.plot(np.random.randn(50).cumsum(), 'c:')
# plt.plot(np.random.randn(50).cumsum(), 'chocolate:') 위에 설정한 색 이외의 색은 불가능. 
# color = 'chocolate', linstyle = '-' 같이 따로 적어줘야 힘
```

- marker
    - 기본 도형 및 점
    
    | 마커코드 | 모양|설명|
    |:---:|:---:|:---:|
    |o|circle|동그라미|
    |.|point|작은 점|
    |s|square|네모|
    |D|diamond|다이아몬드|
    |d|thin diamond|다이아몬드|
    |p|pentagon|오각형|
    |h/H|hexagon|육각형|
    |8|octagon|팍각형|
    |+|plus|플러스기호|
    |x|x|엑스모양|
    |X|X|채워진 x 모양|
    |*|*|별|
    |\||\||vline|
    |_|hline|가로선|


```python
plt.plot(np.random.randn(50).cumsum(), 'b-o')
plt.plot(np.random.randn(50).cumsum(), 'g*')
plt.plot(np.random.randn(50).cumsum(), 'm--p')
plt.plot(np.random.randn(50).cumsum(), 'y-D')
plt.plot(np.random.randn(50).cumsum(), 'k-x', markeredgewidth=3)
# +, x, |, _가 잘 안보이면 너무 얇아서 안보이는 것이다.
# markeredgewidth=3 처럼 두께 조절
```
