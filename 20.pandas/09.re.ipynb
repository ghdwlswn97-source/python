#### Regular Expression
: 파이썬 정규표현식(Regex)을 다루는 내장 라이브러리

```python
import re
```

```python
# one 문자가 있는지 확인
pattern = 'one'
str1 = 'one12345two'
re.match(pattern, str1)
```

```python
match1=re.match(pattern, str1)
match1.group()
```

```python
p= re.compile(r'(\w+)\s+\d+-\d+')        # w -> 영문자, d => 숫자
```

```python
result = p.search('tjoeun 02-1234-1234')
result.group(1)
```

```python
p= r'(\w+)\s+(\d+)-(\d+)'
text = 'tjoeun 02-1234-1234'
print(re.findall(p, text))

p= r'\w+\s+\d+-\d+'
text = 'tjoeun 02-1234-1234'
print(re.findall(p, text))
```
